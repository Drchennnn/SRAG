Two illustrative views; read their axes from the saved figures.

```
from io import BytesIO
from IPython.display import Image as NotebookImage, display
from PIL import Image, ImageDraw

def plot(label):
    image = Image.new("RGB", (440, 260), "white")
    draw = ImageDraw.Draw(image)
    draw.line([(48, 30), (48, 195), (409, 195)], fill="#253744", width=3)
    draw.line([(58, 177), (130, 175), (190, 92), (255, 100), (315, 55), (395, 74)],
              fill="#c45115", width=4)
    draw.text((228, 226), label, fill="#253744", anchor="mm", font_size=24)
    stream = BytesIO()
    image.save(stream, format="PNG")
    return stream.getvalue()

print("Two saved views")
display(NotebookImage(data=plot("Time [ms]")))
display(NotebookImage(data=plot("Frequency [kHz]")))
```
